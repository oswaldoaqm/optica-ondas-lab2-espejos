# Laboratorio 2: los cuatro casos del espejo cóncavo

Datos del Cuadro 2 (apéndice C). Para cada caso calculamos con f = 20 cm (R = 40 cm) la distancia imagen y el aumento teóricos, y los comparamos con lo medido en el riel.

- dᵢ = d₀·f / (d₀ − f)
- M = −dᵢ / d₀
- diferencia porcentual = |dᵢ medido − dᵢ teórico| / dᵢ teórico × 100

```python
from pathlib import Path

import numpy as np
import pandas as pd

DATOS = Path("../datos")
F = 20.0  # cm, focal del fabricante
R = 2 * F  # cm
```

```python
c2 = pd.read_csv(DATOS / "cuadro2_casos.csv")

c2["di_teo_cm"] = c2["d0_cm"] * F / (c2["d0_cm"] - F)
c2["M_teo"] = -c2["di_teo_cm"] / c2["d0_cm"]
c2["dif_di_%"] = (c2["di_exp_cm"] - c2["di_teo_cm"]).abs() / c2["di_teo_cm"].abs() * 100
c2.round(2)
```

## Características de cada imagen

El signo de dᵢ dice si la imagen es real (positivo, delante del espejo) o virtual (negativo, detrás). El signo de M dice si es derecha o invertida, y |M| si es ampliada o reducida.

```python
def describir(di, M):
    tipo = "real" if di > 0 else "virtual"
    orientacion = "derecha" if M > 0 else "invertida"
    if np.isclose(abs(M), 1, atol=0.005):
        tamano = "mismo tamaño"
    elif abs(M) > 1:
        tamano = "ampliada"
    else:
        tamano = "reducida"
    return f"{tipo}, {orientacion}, {tamano}"


c2["imagen_teorica"] = [describir(di, M) for di, M in zip(c2["di_teo_cm"], c2["M_teo"])]
c2[["caso", "d0_cm", "di_exp_cm", "di_teo_cm", "M_estimado", "M_teo", "dif_di_%", "imagen_teorica"]].round(2)
```

## Lectura de la tabla

En los tres casos con imagen real el signo y el tamaño estimados a ojo coinciden con la teoría: reducida para d₀ = 50 cm, del mismo tamaño en d₀ = R = 40 cm y ampliada para d₀ = 35 cm. Las diferencias en dᵢ son de 1.0 %, 0 % y 10.0 %.

La diferencia más grande está en f < d₀ < R. Ahí la imagen se forma a 46.7 cm del espejo, más allá de la fuente, y para atraparla hubo que inclinar el espejo; eso hace más difícil encontrar el punto de mejor enfoque.

En d₀ < f el dᵢ teórico es negativo: la imagen es virtual y no se puede proyectar en la pantalla. Por eso esa fila no tiene dᵢ medido ni diferencia porcentual. La única comprobación posible es mirar dentro del espejo, donde la flecha se ve derecha y ampliada.
